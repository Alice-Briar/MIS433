### Tidying the data

Next, you will further tidy the two DataFrames **stars** and **genres**.

Let's start with **stars**.

```python
# Transform the DataFrame "stars" to a new DataFrame named "stars_long" with the following four columns:
# Series_Title, Released_Year, StarNo (e.g., Star1, Star2, ...), StarName
# Hint: use melt()
stars_long = stars.melt(
    id_vars=['Series_Title', 'Released_Year'],
    var_name='StarNo',
    value_name='StarName'
)

stars_long
```

```python
# Can you transform the DataFrame 'stars_long' back to its original shape?
# Hint: use pivot()
stars_long.pivot(
    index=['Series_Title', 'Released_Year'],
    columns='StarNo',
    values='StarName'
).reset_index()
```

```python
# In the DataFrame "stars_long", find all movies that star "Morgan Freeman".
stars_long[stars_long['StarName'] == 'Morgan Freeman']
```

```python
# Who appeared in Star2 the most times? List the top five actors.
stars_long[stars_long['StarNo'] == 'Star2']['StarName'].value_counts().head(5)
```

```python
# Who stars in the most movies in this list? List the top 20 actors.
stars_long['StarName'].value_counts().head(20)
```

```python
# Which movie stars had the highest total gross in this movie list? Show the top 10 actors.
# Hint: Join "stars_long" and "imdb"; then group by StarName
actor_gross = stars_long.merge(
    imdb[['Series_Title', 'Gross']],
    on='Series_Title'
)

actor_gross.groupby('StarName')['Gross'].sum().sort_values(
    ascending=False
).head(10)
```

```python
# Find the best director-actor duos
# i.e., director-actor pairs that collaborated in at least five movies,
# sort them in descending order of average IMDB_Rating
# Hint: Join imdb and stars_long; group by ['Director','StarName']
duos = imdb.merge(stars_long, on=['Series_Title', 'Released_Year'])

duos.groupby(['Director', 'StarName']).agg(
    Movies=('Series_Title', 'count'),
    Average_Rating=('IMDB_Rating', 'mean')
).query('Movies >= 5').sort_values(
    'Average_Rating', ascending=False
)
```

```python
# Bonus question
# Who did "Amy Adams" co-star with in this movie list?
# Hint: Join stars_long with itself to find pairs of co-stars
amy = stars_long[stars_long['StarName'] == 'Amy Adams']

co_stars = stars_long.merge(amy, on='Series_Title')

co_stars[co_stars['StarName_x'] != 'Amy Adams']['StarName_x'].value_counts()
```

Next, let's reshape the DataFrame **genres**, which is a little more complex.

```python
# Step 1: Split the 'Genre' string by ', ' into a list of individual genres and expand them into separate columns
genres_split = genres['Genre'].str.split(', ', expand=True)
genres_split
```

```python
# Step 2: Rename the columns as: Genre1, Genre2, Genre3
genres_split.columns = ['Genre1', 'Genre2', 'Genre3']
```

```python
# Step 3: Combine ['Series_Title','Released_Year'] from 'genres' with ['Genre1','Genre2','Genre3'] from 'genres_split'.
# Save it to a new DataFrame named 'genres_wide'.
# Hint: Use pd.concat(...)
genres_wide = pd.concat(
    [genres[['Series_Title', 'Released_Year']], genres_split],
    axis=1
)

genres_wide
```

```python
# Step 4: Transform genres_wide to a new DataFrame genres_long with the following four columns:
# Series_Title, Released_Year, GenreNo (e.g., Genre1, Genre2, Genre3), GenreName
# Hint: use melt()
genres_long = genres_wide.melt(
    id_vars=['Series_Title', 'Released_Year'],
    var_name='GenreNo',
    value_name='GenreName'
)

genres_long = genres_long.dropna()

genres_long
```

```python
# How many unique genres (atomic values, e.g., Drama, Comedy, ...) are there?
# How many movies are there for each genre?
print(genres_long['GenreName'].nunique())

genres_long['GenreName'].value_counts()
```

```python
# What is the average IMDB rating for each genre?
# Sort the genres in descending order of average IMDB_Rating.
# Hint: join imdb with genres_long; group by GenreName
genre_ratings = genres_long.merge(
    imdb[['Series_Title', 'IMDB_Rating']],
    on='Series_Title'
)

genre_ratings.groupby('GenreName')['IMDB_Rating'].mean().sort_values(
    ascending=False
)
```

```python
# Who is the "King of Comedy" (i.e., the actor who starred in the most comedy movies)?
# Hint: find all comedies in genres_long ; join with stars_long; group by StarName
comedy = genres_long[genres_long['GenreName'] == 'Comedy']

comedy_stars = comedy.merge(stars_long, on='Series_Title')

comedy_stars['StarName'].value_counts().head(10)
```

```python
# Who are the best action stars?
# i.e., the actors who star in at least 5 action movies, sorted by average IMDB rating in descending order
# Hint: You need to join columns from all three DataFrames: imdb, stars_long, genres_long
action = genres_long[genres_long['GenreName'] == 'Action']
action_stars = action.merge(stars_long, on='Series_Title')
action_stars = action_stars.merge(imdb[['Series_Title', 'IMDB_Rating']], on='Series_Title')

action_stars.groupby('StarName').agg(
    Movies=('Series_Title', 'nunique'),
    Average_Rating=('IMDB_Rating', 'mean')
).query('Movies >= 5').sort_values('Average_Rating', ascending=False)
```

```python
# Create a pivot table to show the number of movies of different genres in different decades.
# Row: GenreName
# Column: Decade
# Hint: https://pandas.pydata.org/docs/reference/api/pandas.pivot_table.html
genre_decades = genres_long.merge(
    imdb[['Series_Title', 'Decade']],
    on='Series_Title'
)

pd.pivot_table(
    genre_decades,
    index='GenreName',
    columns='Decade',
    values='Series_Title',
    aggfunc='count',
    fill_value=0
)
```
